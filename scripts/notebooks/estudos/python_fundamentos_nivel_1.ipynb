#### 🎨 f-strings — formatação moderna
A forma mais limpa de juntar texto com variáveis. Use `f` antes das aspas e coloque variáveis entre `{}`.

## 🔢 Operadores

#### ➕ Aritméticos

```python
a = 10
b = 3

print(a + b)    # soma           -> 13
print(a - b)    # subtração      -> 7
print(a * b)    # multiplicação  -> 30
print(a / b)    # divisão        -> 3.333...
print(a // b)   # divisão inteira -> 3
print(a % b)    # resto          -> 1
print(a ** b)   # potência       -> 1000 (10 elevado a 3)
```

**Dica prática:** o operador `%` (resto) é muito usado para descobrir se um número é par/ímpar, múltiplo de algo, ou ciclar valores.
- Par: `numero % 2 == 0`
- Múltiplo de 5: `numero % 5 == 0`

#### ⚖️ Comparação

```python
print(10 > 5)    # maior         -> True
print(10 < 5)    # menor         -> False
print(10 >= 10)  # maior ou igual -> True
print(10 <= 9)   # menor ou igual -> False
print(10 == 5)   # igual         -> False
print(10 != 5)   # diferente     -> True
```

**Atenção:** `=` ATRIBUI valor (coloca na variável). `==` COMPARA se dois valores são iguais. Confundir os dois é o erro mais comum em programação.

#### 🔗 Lógicos

```python
print(True and False)   # E   -> False (precisa dos dois)
print(True or False)    # OU  -> True (basta um)
print(not True)         # NÃO -> False (inverte)
```

**Exemplo prático combinando os três:**

```python
idade = 25
tem_cnh = False
tem_carro = True

# Pode dirigir? (maior de 18 E tem CNH)
pode_dirigir = idade >= 18 or tem_cnh
print(f"Pode dirigir: {pode_dirigir}")

# Precisa de transporte? (não tem carro OU não tem CNH)
precisa_transporte = not tem_carro or not tem_cnh
print(f"Precisa de transporte: {precisa_transporte}")
```

#### 🔄 for

```python
# range(n) -> de 0 até n-1
for i in range(5):
    print(i)

```

**range com início, fim e passo:**

```python
for i in range(5):
    print(i)

# range(início, fim) -> de início até fim-1
for i in range(1, 6):
    print(i)   # 1, 2, 3, 4, 5

print("---")

# range(início, fim, passo) -> pula de passo em passo
for i in range(0, 20, 5):
    print(i)   # 0, 5, 10, 15

print("---")

# range com passo negativo -> contagem regressiva
for i in range(10, 0, -1):
    print(i)   # 10, 9, 8, ..., 1
```

## 📚 Estruturas de Dados


#### 📋 Listas — criar, acessar, alterar

```python
numeros = [1, 2, 3, 4]
print(numeros)

# Acessar por índice (começa em 0)
print(numeros[0])    # primeiro -> 1
print(numeros[2])    # terceiro -> 3
print(numeros[-1])   # último   -> 4

# Tamanho da lista
print(len(numeros))  # 4
```

**Alterando itens:**

```python
frutas = ["maçã", "banana", "uva"]
print(frutas[1])   # banana

# Alterar item por índice
frutas[1] = "laranja"
print(frutas)   # ['maçã', 'laranja', 'uva']
```

**Adicionando itens:**

```python
numeros = [1, 2, 3]

# append() -> adiciona UM no FINAL
numeros.append(4)
print(numeros)   # [1, 2, 3, 4]

# insert(posição, valor) -> adiciona em posição específica
numeros.insert(0, 0)
print(numeros)   # [0, 1, 2, 3, 4]

# extend() -> adiciona vários de uma vez
numeros.extend([5, 6, 7])
print(numeros)   # [0, 1, 2, 3, 4, 5, 6, 7]
```

**Removendo itens:**

```python
frutas = ["maçã", "banana", "uva", "banana"]

# remove() -> remove a PRIMEIRA ocorrência do valor
frutas.remove("banana")
print(frutas)   # ['maçã', 'uva', 'banana']

# pop() -> remove e retorna o item de uma posição (último por padrão)
ultimo = frutas.pop()
print(f"Removido: {ultimo}")
print(frutas)  # ['maçã', 'uva']

# pop(índice) -> remove de posição específica
primeiro = frutas.pop(0)
print(f"Removido: {primeiro}")
print(frutas)
```

**Fatiamento (slicing) — pegar um pedaço da lista:**

```python
numeros = [10, 20, 30, 40, 50, 60]

print(numeros[1:4])    # do índice 1 ao 3 -> [20, 30, 40]

print(numeros[:3])     # do início até o 2 -> [10, 20, 30]

print(numeros[3:])     # do índice 3 até o fim -> [40, 50, 60]

print(numeros[::2])    # de 2 em 2 -> [10, 30, 50]

print(numeros[::-1])   # invertida -> [60, 50, 40, 30, 20, 10]
```

**Outras operações úteis:**

```python
numeros = [3, 1, 4, 1, 5, 9, 2, 6]

# print(sum(numeros))    # soma -> 31
# print(max(numeros))    # maior -> 9
# print(min(numeros))    # menor -> 1
print(sorted(numeros)) # ordenada (não altera original)
print(numeros)         # original intacta


numeros.sort()       # ordena IN-PLACE (altera original)
print(numeros)

frutas = ["maçã", "banana", "uva"]

# Verificar se um valor está na lista
print(5 in numeros)    # True
print(100 in numeros)  # False
print("maçã" in frutas)  # True
```

**Desempacotamento (unpacking) — uso mais comum de tuplas:**
Permite atribuir os valores de uma tupla a várias variáveis de uma vez.

```python
coordenada = (10, 20)

x, y = coordenada
print(f"x={x}, y={y}") # x=10, y=20

#-------------------------

# Funciona para trocar valores sem variável auxiliar
a = 1
b = 2

a, b = b, a
1,2 = 2,1

print(f"a={a}, b={b}")   # a=2, b=1

#-------------------------

# E para receber múltiplos retornos de uma função
def min_max(numeros):
    return min(numeros), max(numeros)

menor, maior = min_max([3, 1, 4, 1, 5, 9])
print(f"menor={menor}, maior={maior}")
```

**Iterando sobre dicionários:**

```python
pessoa = {
    "nome": "Vinicius",
    "idade": 33,
    "cidade": "São Paulo",
    "tipo_sanguineo": "O+"}

# .keys() -> só as chaves
for chave in pessoa.keys():
    print(chave)    # nome, idade, cidade

print("---")

# .values() -> só os valores
for valor in pessoa.values():
    print(valor)    # Vinicius, 33, São Paulo

print("---")

# .items() -> chave E valor juntos (mais usado)
for chave, valor in pessoa.items():
    print(f"{chave}: {valor}")
```

**Uso prático — operações entre conjuntos:**

```python
time_a = {"Ana", "Bruno", "Carla", "Diego"}
time_b = {"Carla", "Diego", "Eduardo", "Fernanda"}

# Quem está nos DOIS times (interseção)
print(time_a & time_b)   # {'Carla', 'Diego'}

# Todos de qualquer time (união)
print(time_a | time_b)

# Quem está SÓ no time A (diferença)
print(time_a - time_b)   # {'Ana', 'Bruno'}
print(time_b - time_a)   # {'Eduardo', 'Fernanda'}

# Remover duplicatas de uma lista
lista_com_repetidos = [1, 2, 2, 3, 3, 3, 4]

unicos = list(set(lista_com_repetidos))
unicos = list(set([1,2,3,4]))
unicos = list({1,2,3,4})
unicos = [1,2,3,4]

print(unicos)   # [1, 2, 3, 4]
```

#### 🟢 Exercício 1 — Par ou Ímpar

#### 🔵 Exercício 3 — Soma até N

Implemente uma função que receba um número inteiro n como parâmetro e retorne a soma de todos os números de 1 até n. 
Por exemplo, se o valor de entrada for 5, o resultado esperado é 15, pois corresponde à soma de 1 + 2 + 3 + 4 + 5. 
Utilize um laço de repetição (for) para percorrer os valores e acumular o resultado. 
Lembre-se de que a função deve retornar o valor final, e não apenas imprimi-lo.

```python
# SEU CÓDIGO AQUI
def soma_ate_n(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total


print(soma_ate_n(5))    # 15
print(soma_ate_n(10))   # 55
```

#### 🔴 Exercício 4 — Distribuição de Notas

Crie um programa que simule a distribuição de notas em um saque de caixa eletrônico. 
O programa deve receber um valor inteiro informado pelo usuário e calcular a quantidade mínima de notas necessárias para compor esse valor, 
considerando as seguintes cédulas disponíveis: 100, 50, 20, 10, 5, 2 e 1. Ao final, o programa deve exibir quantas notas de cada valor serão utilizadas. 
Para resolver o problema, utilize divisão inteira (//) para calcular a quantidade de notas e o operador de resto (%) para atualizar o valor restante.

```python
# SEU CÓDIGO AQUI
valor = int(input("Digite o valor do saque: R$ "))

notas = [100, 50, 20, 10, 5, 2, 1]
distribuicao = {}

for nota in notas:
    quantidade = valor // nota   # quantas notas desse valor cabem
    if quantidade > 0:
        distribuicao[nota] = quantidade
    valor = valor % nota         # sobra para a próxima nota

for nota, quantidade in distribuicao.items():
    print(f"{quantidade} nota(s) de R$ {nota}")
```

# 💻 Miniprojeto — Caixa Eletrônico


Neste miniprojeto, você deverá desenvolver um simulador completo de caixa eletrônico, aplicando os principais conceitos aprendidos até aqui. 
O programa deve solicitar ao usuário um valor de saque e validar essa entrada, garantindo que seja um número inteiro e maior que zero. 
Em seguida, deve calcular a quantidade de notas necessárias para compor o valor informado, utilizando as cédulas disponíveis (100, 50, 20, 10, 5, 2 e 1). 
Por fim, o programa deve exibir o resultado de forma organizada e legível, informando quantas notas de cada tipo serão entregues. 
Como desafio adicional, você pode implementar melhorias como tratamento de erros com try/except, repetição do programa até que o usuário decida sair e formatação mais elegante da saída.

```python
def calcular_notas(valor):
    notas = [100, 50, 20, 10, 5, 2, 1]
    resultado = {}

    for nota in notas:
        quantidade = valor // nota
        if quantidade > 0:
            resultado[nota] = quantidade
        valor %= nota

    return resultado


print("=" * 50)
print("        CAIXA ELETRÔNICO PYTHON")
print("=" * 50)

while True:
    try:
        valor = int(input("\nDigite o valor do saque: R$ "))

        if valor <= 0:
            print("❌ O valor deve ser maior que zero.")
            continue

        notas = calcular_notas(valor)

        print("\n" + "=" * 50)
        print(f"Valor solicitado: R$ {valor}")
        print("-" * 50)
        print("Notas entregues:")

        for nota, quantidade in notas.items():
            print(f"R$ {nota:>3} -> {quantidade} nota(s)")

        print("=" * 50)

    except ValueError:
        print("❌ Entrada inválida! Digite apenas números inteiros.")
        continue

    opcao = input("\nDeseja realizar outro saque? (S/N): ").strip().upper()

    if opcao != "S":
        print("\nObrigado por utilizar nosso caixa eletrônico!")
        break
```
