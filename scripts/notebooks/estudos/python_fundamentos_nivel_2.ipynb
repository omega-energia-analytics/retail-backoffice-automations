## 🧠 Introdução

No Nível 1 vimos listas e dicionários — ótimos para poucos dados. Mas quando o
assunto é **planilha de dados** (faturas, clientes, tickets...) com milhares de
linhas e várias colunas, a ferramenta certa é a biblioteca **`pandas`**.

O `pandas` gira em torno de duas estruturas:
- **`Series`** → uma **coluna** (uma sequência de valores com rótulos).
- **`DataFrame`** → uma **tabela** inteira (várias `Series` lado a lado).

Pense num `DataFrame` como uma aba de Excel dentro do Python: linhas, colunas,
cabeçalho — só que muito mais rápido e programável.

## 🧵 Series — a peça básica (uma coluna)

Antes da tabela, a coluna. Uma `Series` é uma sequência de valores com um
**índice** (os rótulos à esquerda). Se você não definir um índice, o pandas usa
0, 1, 2, ...

```python
import pandas as pd


consumo = pd.Series([320, 145, 500, 90])
print(consumo)

# Repare: à ESQUERDA vem o índice (0,1,2,3) e à DIREITA os valores.
# O dtype int64 no rodapé é o tipo dos dados da Series.
```

```python
import pandas as pd


# Podemos dar rótulos ao índice (vira quase um dicionário "turbinado")
consumo = pd.Series([320, 145, 500, 90], index=["Ana", "Bruno", "Carla", "Diego"],)
# print(consumo)
# print("---")

# # Acessar pelo rótulo
# print(consumo["Carla"])   # 500


# # Series entende operações em bloco (vetorizado) — sem precisar de for!
print(consumo * 1.05)     # reajuste de 5% em TODOS de uma vez
```

> 💡 **A grande sacada do pandas:** operações acontecem na **coluna inteira de
> uma vez** (isso se chama *vetorização*). Você quase nunca vai precisar de um
> `for` para percorrer linha a linha — e o código fica mais curto e muito mais
> rápido.

## 📍 Selecionando linhas — `loc` e `iloc`

- **`iloc`** → seleciona por **posição** (o número da linha, começando em 0).
- **`loc`**  → seleciona por **rótulo** (o valor do índice).

Como nosso índice é 0,1,2,... os dois parecem iguais agora — mas a diferença
fica clara quando o índice vira, por exemplo, o nome do cliente.

```python
print(faturas)
```

```python
print(faturas.iloc[0])      # primeira linha (posição 0) -> vem como Series

# print("-" * 40)

# print(faturas.iloc[-1])     # última linha
```

```python
# iloc também aceita fatias (slicing), igual às listas do Nível 1
faturas.iloc[0:3]           # linhas nas posições 0, 1 e 2
```

```python
# [linhas, colunas] -> as 3 primeiras linhas, só cliente e valor
faturas.iloc[
     0:3, 
    [0, 4]
]
```

```python
print(faturas)
```

```python
# loc trabalha por RÓTULO. Nosso índice ainda é 0,1,2,... — então
# faturas.loc["Ana Souza", "valor"] dá KeyError: não existe a LINHA "Ana Souza".

# Para acessar pelo nome, primeiro colocamos o cliente no índice:

por_cliente = faturas.set_index("cliente")   # cópia com o índice trocado

# print(por_cliente.index)
# print("-" * 30)

# print(por_cliente)

# # # Agora sim: loc[linha, coluna] pelos rótulos
# print(por_cliente.loc["Ana Souza", "valor"])                     # 287.4
# print("-" * 30)
# print(por_cliente.loc["Ana Souza", ["consumo_kwh", "valor"]])    # várias colunas
# print("-" * 30)
print(por_cliente.loc["Ana Souza":"Bruno Lima", "valor"])        # fatia por rótulo

# # por_cliente.reset_index()   # desfaz: cliente volta a ser coluna comum
```

> ⚠️ **Duas pegadinhas do `loc`:**
> 1. **A fatia do `loc` INCLUI o fim.** `iloc[0:3]` traz 3 linhas (para *antes*
>    da posição 3), mas `loc["Ana Souza":"Bruno Lima"]` traz as duas pontas.
>    Fatia por posição é exclusiva; por rótulo é inclusiva.
> 2. **Guarde o `set_index` numa variável nova** (`por_cliente = ...`) em vez de
>    sobrescrever `faturas`. Se você trocar o índice da tabela original, a coluna
>    `cliente` deixa de existir e toda célula seguinte que faz
>    `faturas["cliente"]` passa a dar `KeyError`.

## 🔬 Filtrando linhas (a operação mais importante!)

Filtrar é responder perguntas do tipo *"quais faturas estão vencidas?"*. Funciona
em 2 passos:
1. Você cria uma **condição**, que vira uma Series de `True`/`False`.
2. Você passa essa condição pra dentro dos colchetes; o pandas devolve só as
   linhas onde deu `True`.

```python
import pandas as pd

faturas = [ 
    {
    "cliente": "Ana Souza",  
    "distribuidora": "Coelba",
    "mes_ref": "2026-01",
    "consumo_kwh": 320,
     "valor": 287.40,
     "status": "pago"
     },
    {
    "cliente": "Bruno Lima", "distribuidora": "CPFL", "mes_ref": "2026-01", "consumo_kwh": 145, "valor": 132.10, "status": "pago"
     },
    {
    "cliente": "Carla Santos", "distribuidora": "Energisa", "mes_ref": "2026-01", "consumo_kwh": 210, "valor": 189.70, "status": "pendente"
     },
    {
    "cliente": "David Costa", "distribuidora": "Light", "mes_ref": "2026-01", "consumo_kwh": 180,"valor": 162.00, "status": "pago"
    },
    {
    "cliente": "Elena Dias", "distribuidora": "Coelba", "mes_ref": "2026-01", "consumo_kwh": 250, "valor": 225.00, "status": "pendente"
    }
]

faturas = pd.DataFrame(faturas)
```

```python
faturas
```

```python
# Passo 1: a condição sozinha é uma máscara de True/False
faturas["valor"] < 287.4
```

```python
# Passo 2: usar a máscara para filtrar -> só as faturas acima de R$ 300
faturas[faturas["valor"] > 200]

# faturas = faturas[faturas["valor"] > 200]
```

```python
faturas
```

```python
# Filtro por igualdade de texto
# faturas[faturas["status"] == "pago"]

faturas[faturas["cliente"] == "Carla Santos"]
```

#### 🔗 Combinando condições

Para juntar condições use `&` (E), `|` (OU) e `~` (NÃO). E **cada condição
precisa estar entre parênteses**.

```python
# Faturas da Coelba E com consumo acima de 400 kWh
faturas[
    (faturas["distribuidora"] == "Coelba")
    & 
    (faturas["consumo_kwh"] > 200)
    ]
```

```python
# Faturas vencidas OU em aberto (ou seja, não pagas)
faturas[
    (faturas["status"] == "pago") 
    | 
    (faturas["status"] == "em aberto")]
```

> ⚠️ **O erro nº 1 de quem vem do Nível 1:** usar `and`/`or` aqui. NÃO funciona
> com colunas do pandas — dá `ValueError`. Dentro de um filtro é sempre `&` e
> `|`, e **cada condição entre parênteses** (senão a precedência dos operadores
> quebra tudo).

```python
# Ferramentas que deixam o filtro mais limpo:

# .isin() -> "está em uma lista de valores?" (evita vários | encadeados)
faturas[
    faturas["distribuidora"].isin(["Coelba", "Aegea"])
    ].sample(1)

# faturas[
#     (faturas["distribuidora"] == "Coelba") & (faturas["distribuidora"] == "Aegea")
#     ]
```

```python
# .between() -> intervalo (inclusivo nas duas pontas)
cliente_baixo = faturas[faturas["valor"].between(50, 100)]
cliente_medio = faturas[faturas["valor"].between(100, 200)]
cliente_alto = faturas[faturas["valor"].between(200, 300)]

print("cliente baixo:", len(cliente_baixo))
print("cliente médio:", len(cliente_medio))
print("cliente alto:", len(cliente_alto))
```

## ➕ Criando e alterando colunas

Criar coluna nova é só atribuir a `df["nova_coluna"]`. As operações são
**vetorizadas**: valem para a coluna inteira de uma vez.

```python
import pandas as pd
import numpy as np

faturas = [ 
    {
    "cliente": "Ana Souza",  
    "distribuidora": "Coelba",
    "mes_ref": "2026-01",
    "consumo_kwh": 320,
     "valor": 287.40,
     "status": "pago"
     },
    {
    "cliente": "Bruno Lima", "distribuidora": "CPFL", "mes_ref": "2026-01", "consumo_kwh": 145, "valor": 132.10, "status": "pago"
     },
    {
    "cliente": "Carla Santos", "distribuidora": "Energisa", "mes_ref": "2026-01", "consumo_kwh": 210, "valor": 189.70, "status": "pendente"
     },
    {
    "cliente": "David Costa", "distribuidora": "Light", "mes_ref": "2026-01", "consumo_kwh": 180,"valor": 162.00, "status": "pago"
    },
    {
    "cliente": "Elena Dias", "distribuidora": "Coelba", "mes_ref": "2026-01", "consumo_kwh": 250, "valor": 225.00, "status": "pendente"
    }
]

faturas = pd.DataFrame(faturas)
```

```python
# Preço médio por kWh de cada fatura (coluna nova a partir de duas existentes)
faturas["preco_kwh"] = faturas["valor"] / faturas["consumo_kwh"]

faturas[["cliente", "valor", "consumo_kwh", "preco_kwh"]].head()
```

```python
# Arredondando a coluna nova para 2 casas
faturas["preco_kwh"] = faturas["preco_kwh"].round(2)
faturas[["cliente", "preco_kwh"]].head()
```

#### 🔀 Coluna condicional com `np.where`

`np.where(condição, valor_se_verdadeiro, valor_se_falso)` é o "if" vetorizado —
cria uma coluna baseada numa regra, sem `for`.

```python
# Marca cada fatura como "alta" ou "normal" conforme o valor
faturas["faixa"] = np.where(faturas["valor"] > 300, "alta", "normal")

faturas[["cliente", "valor", "faixa"]]
```

#### 🧰 `.str` e `.apply` — transformando os valores

- O acessador **`.str`** aplica operações de texto na coluna toda (`.upper()`,
  `.lower()`, `.strip()`, `.replace()`...).
- **`.apply(funcao)`** roda uma função sua em cada valor — o canivete suíço para
  quando não existe um método pronto.

```python
# .str -> distribuidora em MAIÚSCULAS
faturas["distribuidora"].str.upper()
```

```python
# .apply com uma função nossa: classificar o consumo
def classificar_consumo(kwh):
    if kwh < 150:
        return "baixo"
    elif kwh < 400:
        return "médio"
    else:
        return "alto"

faturas["perfil"] = faturas["consumo_kwh"].apply(classificar_consumo)
faturas[["cliente", "consumo_kwh", "perfil"]]
```

## 🧹 Dados faltantes (`NaN`)

No mundo real quase toda tabela tem buracos: um campo que não veio, uma leitura
que faltou. O pandas representa isso como **`NaN`** (*Not a Number*). Ignorar
esses buracos é uma das maiores fontes de erro em análise de dados.

```python
import pandas as pd
import numpy as np 
# Vamos simular uma tabela com furos (repare nos np.nan)
dados = pd.DataFrame({
    "cliente":     ["Ana", "Bruno", "Carla", "Diego"],
    "consumo_kwh": [320, np.nan, 500, np.nan],
    "valor":       [287.40, 132.10, np.nan, 88.75],
})
dados
```

```python
# Onde estão os buracos? .isna() devolve True onde é NaN
dados.isna()
```

```python
# Quantos NaN por coluna? (truque clássico: .isna().sum())
dados.isna().sum()
```

```python
# Preencher os buracos com um valor -> .fillna()
dados["consumo_kwh"].fillna(0)       # troca NaN por 0
```

```python
# Preencher com a MÉDIA da própria coluna (estratégia comum)
media = dados["valor"].mean()        # a média já IGNORA os NaN automaticamente
media

dados["valor"].fillna(round(media, 2))
```

```python
dados
```

```python

# Descartar as linhas que têm qualquer NaN -> .dropna()
dados.dropna()
```

> ⚠️ **`NaN` é contagioso:** qualquer conta com `NaN` vira `NaN`
> (`10 + NaN = NaN`). Já funções de resumo como `.mean()`, `.sum()`, `.max()`
> **ignoram** o `NaN` por padrão. Decida sempre de forma consciente: preencher
> (`fillna`) ou descartar (`dropna`)?

## 📊 Agrupando e resumindo (`groupby`) — o coração da análise

Aqui é onde o pandas brilha. `groupby` segue a lógica
**dividir → aplicar → juntar**: separa as linhas em grupos, aplica um cálculo em
cada grupo e junta tudo num resumo. É o equivalente ao `GROUP BY` do SQL.

Antes, o atalho mais usado do dia a dia:

```python
faturas
```

```python
# value_counts() -> quantas linhas de cada valor (uma "contagem" rápida)
faturas["distribuidora"].value_counts()
```

```python
# value_counts também dá a proporção
faturas["status"].value_counts(normalize=True)   # fração de cada status
```

#### Agora o `groupby` de verdade

```python
# Valor TOTAL faturado por distribuidora
faturas.groupby("distribuidora")["valor"].sum()
```

```python
# Consumo MÉDIO por distribuidora
faturas.groupby("distribuidora")["consumo_kwh"].mean().round(1)
```

```python
# Quantas faturas por status
faturas.groupby("status")["cliente"].count()
```

#### Várias métricas de uma vez com `.agg()`

`.agg()` deixa você pedir vários cálculos ao mesmo tempo e ainda nomear as
colunas do resultado. É assim que se monta um relatório de verdade.

```python
resumo = faturas.groupby("distribuidora").agg(
    qtd_faturas   = ("cliente", "count"),
    total_faturado= ("valor", "sum"),
    ticket_medio  = ("valor", "mean"),
    consumo_total = ("consumo_kwh", "sum"),
)

resumo.round(2)
```

```python
# Agrupando por MAIS DE UMA coluna: distribuidora x status
faturas.groupby(["distribuidora", "status"])["valor"].sum()
```

## 🔗 Juntando DataFrames

Duas formas de combinar tabelas:
- **`merge`** → junta pelo conteúdo de uma coluna em comum (é o `JOIN` do SQL).
- **`concat`** → empilha tabelas (uma embaixo da outra, ou lado a lado).

```python
# Uma tabela auxiliar com dados de cada distribuidora
distribuidoras = pd.DataFrame({
    "distribuidora": ["Coelba", "CPFL", "Aegea"],
    "estado":        ["BA", "SP", "MT"],
    "regiao":        ["Nordeste", "Sudeste", "Centro-Oeste"],
})
distribuidoras
```

```python
# merge: para cada fatura, "puxa" estado e região pela coluna em comum
faturas_completa = faturas.merge(distribuidoras, on="distribuidora", how="left")

faturas_completa[["cliente", "distribuidora", "estado", "regiao", "valor"]].head()
```

> 🔗 **O parâmetro `how` decide quem sobra:** `left` mantém todas as linhas da
> tabela da esquerda (as faturas) — o mais usado; `inner` só mantém as que dão
> match nas duas; `outer` mantém tudo. É idêntico ao `LEFT JOIN` / `INNER JOIN`
> do SQL.

# 💻 Miniprojeto — Relatório de faturamento e inadimplência

Junte tudo do notebook num pequeno relatório. A partir de uma base de faturas
(pode ser a `faturas` que já temos, ou uma maior que você invente), produza:

1. **Higienização:** garanta que `valor` e `consumo_kwh` são numéricos e trate
   eventuais `NaN` (preencher ou descartar — decida e comente o porquê).
2. **Colunas derivadas:**
   - `preco_kwh` = `valor` / `consumo_kwh` (arredondado).
   - `pago` = coluna booleana (`True`/`False`) indicando se o status é `"pago"`.
3. **Enriquecimento:** traga `estado` e `regiao` via `merge` com `distribuidoras`.
4. **Relatório por distribuidora** (`groupby().agg`): total faturado, ticket
   médio, e **valor inadimplente** (soma do `valor` onde `pago == False`).
5. **Indicador final:** calcule e imprima, com f-string, a **taxa de
   inadimplência geral** = valor não pago / valor total, formatada como `%`.

Desafios extras: exporte o relatório com `to_csv`; ordene o ranking pelo valor
inadimplente; destaque a distribuidora com maior inadimplência.

```python
# SEU CÓDIGO AQUI
distribuidoras = pd.DataFrame({
    "distribuidora": ["Coelba", "CPFL", "Aegea"],
    "estado": ["BA", "SP", "RJ"],
    "regiao": ["Nordeste", "Sudeste", "Sudeste"]
})

# ==============================
# 1. HIGIENIZAÇÃO
# ==============================

# Garante que valor e consumo_kwh sejam numéricos.
# Valores que não puderem ser convertidos viram NaN.
faturas["valor"] = pd.to_numeric(faturas["valor"], errors="coerce")
faturas["consumo_kwh"] = pd.to_numeric(faturas["consumo_kwh"], errors="coerce")

# Remove linhas com NaN em valor ou consumo_kwh.
# Decisão: como esses campos são essenciais para os cálculos,
# é mais seguro descartar as linhas incompletas.
faturas = faturas.dropna(subset=["valor", "consumo_kwh"])


# ==============================
# 2. COLUNAS DERIVADAS
# ==============================

# Calcula o preço médio de cada kWh e arredonda para 2 casas decimais.
faturas["preco_kwh"] = (
    faturas["valor"] / faturas["consumo_kwh"]
).round(2)

# Cria uma coluna booleana indicando se a fatura foi paga.
faturas["pago"] = faturas["status"] == "pago"


# ==============================
# 3. ENRIQUECIMENTO
# ==============================

# Junta as informações de estado e região à tabela de faturas.
faturas = faturas.merge(
    distribuidoras[["distribuidora", "estado", "regiao"]],
    on="distribuidora",
    how="left"
)


# ==============================
# 4. RELATÓRIO POR DISTRIBUIDORA
# ==============================

# Cria um resumo por distribuidora.
relatorio = faturas.groupby("distribuidora").agg(
    total_faturado=("valor", "sum"),              # Soma de todas as faturas
    ticket_medio=("valor", "mean"),               # Valor médio das faturas
    valor_inadimplente=("valor",                 # Soma somente das não pagas
                        lambda x: x[~faturas.loc[x.index, "pago"]].sum())
).round(2)


# Ordena o ranking pela maior inadimplência.
relatorio = relatorio.sort_values(
    "valor_inadimplente",
    ascending=False
)


# ==============================
# 5. TAXA DE INADIMPLÊNCIA GERAL
# ==============================

# Calcula o valor total das faturas.
valor_total = faturas["valor"].sum()

# Calcula quanto ainda não foi pago.
valor_nao_pago = faturas.loc[~faturas["pago"], "valor"].sum()

# Calcula a taxa de inadimplência.
taxa_inadimplencia = valor_nao_pago / valor_total

# Imprime a taxa formatada como porcentagem.
print(f"Taxa de inadimplência geral: {taxa_inadimplencia:.2%}")


# ==============================
# 6. DESTAQUE DA MAIOR INADIMPLÊNCIA
# ==============================

# Identifica a distribuidora com maior valor inadimplente.
maior_inadimplencia = relatorio["valor_inadimplente"].idxmax()

print(
    f"Distribuidora com maior inadimplência: "
    f"{maior_inadimplencia}"
)


# ==============================
# 7. EXPORTAÇÃO
# ==============================

# Exporta o relatório para um arquivo CSV.
relatorio.to_csv(
    "relatorio_inadimplencia.csv"
)

# Exibe o relatório final.
relatorio
```
